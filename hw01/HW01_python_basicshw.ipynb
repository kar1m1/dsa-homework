````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
# 1) Приводим города все к единому виду и сохраняем в новый список
cities_clean = [city.strip().title() for city in cities]
````

````python
# 2) Преобразование строк в числа
amounts_int = [int(amount) for amount in amounts]
print("Города:", cities_clean)
print("Суммы:", amounts_int)
````

````python
# 3) Общее число заказов, общая сумма, средний чек
total_orders = len(amounts_int)
total_sum = sum(amounts_int)
average_check = round(total_sum / total_orders)
print("Общее число заказов:", total_orders)
print("Общая сумма:", total_sum)
print("Средний чек:", average_check)
````

````python
# 4) Неоплаченные заказы
unpaid_orders = sum(1 for status in paid if not status)
unpaid_sum = sum(
    amount for amount, status in zip(amounts_int, paid)
    if not status
)
print("Неоплаченные заказы:", unpaid_orders)
print("Сумма неоплаченных заказов:", unpaid_sum)
````

````python
# 5) Вывод таблицы с использованием zip и f строки
print(f"{'Город':<12} | {'Сумма':>8} | {'Статус'}")
print("-" * 35)
for city, amount, status in zip(cities_clean, amounts_int, paid):
    payment_status = "Оплачен" if status else "Не оплачен"
    print(f"{city:<12} | {amount:>8} | {payment_status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# 6) Оплаченные выручки по Алмате и Астане
almaty_revenue = sum(
    amount for city, amount, status in zip(cities_clean, amounts_int, paid)
    if city == "Almaty" and status
)
astana_revenue = sum(
    amount for city, amount, status in zip(cities_clean, amounts_int, paid)
    if city == "Astana" and status
)
print("Выручка Алматы:", almaty_revenue, "₸")
print("Выручка Астаны:", astana_revenue, "₸")
````

````python
# 7) Индек и город самого крупного оплаченного заказа
max_amount = -1
max_index = -1
max_city = ""
for i in range(len(amounts_int)):
    if paid[i] and amounts_int[i] > max_amount:
        max_amount = amounts_int[i]
        max_index = i
        max_city = cities_clean[i]
print("Индекс:", max_index)
print("Город:", max_city)
print("Сумма:", max_amount, "₸")
````

````python
# 8) Доля заказа в сб и вс в процентах
weekend_orders = sum(1 for day in days if day in ["сб", "вс"])
weekend_percent = weekend_orders / len(days) * 100
print("Заказов в выходные:", weekend_orders)
print("Доля выходных заказов:", round(weekend_percent, 2), "%")
````

````python
# 9) Задача копилка
target = 400000
monthly_saving = 15000
total = 0
months = 0
while total < target:
    total += monthly_saving
    months += 1
print("Месяцев:", months)
print("Накоплено:", total, "₸")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490

```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
# 10) Оплаченная выручка по каждому городу
unique_cities = []
for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)
revenue_by_city = []

for city in unique_cities:
    revenue = 0
    for c, amount, status in zip(cities_clean, amounts_int, paid):
        if c == city and status:
            revenue += amount
    revenue_by_city.append(revenue)
for i in range(len(revenue_by_city)):
    for j in range(i + 1, len(revenue_by_city)):
        if revenue_by_city[j] > revenue_by_city[i]:
            revenue_by_city[i], revenue_by_city[j] = revenue_by_city[j], revenue_by_city[i]
            unique_cities[i], unique_cities[j] = unique_cities[j], unique_cities[i]

print("Города:", unique_cities)
print("Выручка:", revenue_by_city)
for city, revenue in zip(unique_cities, revenue_by_city):
    print(f"{city:<12} | {revenue:>8} ₸")
````

````python
# 11) Медиана сумма заказа
amounts_sorted = amounts_int.copy()
amounts_sorted.sort()
n = len(amounts_sorted)

if n % 2 == 0:
    median = (amounts_sorted[n // 2 - 1] + amounts_sorted[n // 2]) / 2
else:
    median = amounts_sorted[n // 2]
print("Отсортированные суммы:", amounts_sorted)
print("Медиана:", median)
````

````python
# 12) Гистограмма выручки по дням
unique_days = []
for day in days:
    if day not in unique_days:
        unique_days.append(day)
revenue_by_day = []

for day in unique_days:
    revenue = 0
    for d, amount, status in zip(days, amounts_int, paid):
        if d == day and status:
            revenue += amount
    revenue_by_day.append(revenue)
for day, revenue in zip(unique_days, revenue_by_day):
    bar = "█" * (revenue // 5000)
    print(f"{day} | {bar:<10} {revenue}")
````
